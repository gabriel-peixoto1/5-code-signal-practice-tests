Implement the following operations:

* add_product(product_id, name, price, stock)

- Adds a new product to the system.

* get_product(product_id)

- Returns the product associated with the given ID.
- If the product does not exist, return None.

* update_stock(product_id, quantity)

- Changes the stock of an existing product by adding quantity to its current stock.
