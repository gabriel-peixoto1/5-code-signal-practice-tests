```python
import subprocess
import sys
import re

# ===================== NÍVEL 01 ===================================

result = subprocess.run(
    [sys.executable, "-m", "unittest", "tests.test_level1"],
    capture_output=True,
    text=True
)

output = result.stdout + result.stderr

tests = re.search(r"Ran (\d+) tests?", output)
failures = re.search(r"failures=(\d+)", output)
errors = re.search(r"errors=(\d+)", output)

total = int(tests.group(1)) if tests else 0
failed = int(failures.group(1)) if failures else 0
failed += int(errors.group(1)) if errors else 0

passed = total - failed

print("=" * 50)
print("LEVEL 1 — USUÁRIOS E EVENTOS")
print("=" * 50)
print()

if failed == 0:
    print(f"✓ {passed}/{total} testes passaram")
    print()
    print("STATUS: APROVADO")
else:
    print(f"✓ {passed}/{total} testes passaram")
    print(f"✗ {failed} teste(s) falharam")
    print()
    print("STATUS: FALHOU")
```
