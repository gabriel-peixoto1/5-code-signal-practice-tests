Level 1 — Customers and Orders

Implement the basic customer and delivery order management functionality.

Each customer has:

- customer_id
- name

Each order has:

- order_id
- customer_id
- description
- weight
- status

A newly created order must have status "pending".

add_customer

- If a customer with the same customer_id already exists, return False.
- Otherwise, create the customer and return True.

add_order

- Create a new order with status "pending".

Return False if:

- an order with the same order_id already exists;
- the specified customer does not exist;
- weight is less than or equal to 0.

Otherwise:

- create the order;
- set its status to "pending";
- return True.

get_order

- Return the order with the specified ID.
-Return None if the order does not exist.

```python
class DeliveryManager:
    def add_customer(self, customer_id: int, name: str) -> bool:
        pass

    def add_order(
        self,
        order_id: int,
        customer_id: int,
        description: str,
        weight: float
    ) -> bool:
        pass

    def get_order(self, order_id: int) -> dict | None:
        pass

class DeliveryManagerImpl(DeliveryManager):

    def __init__(self):
        self.customers = []
        self.orders = [] 
        self.drivers = []

    def add_customer(self, customer_id: int, name: str) -> bool:
        
        for customer in self.customers: 
            if customer['customer_id'] == customer_id:
                return False

        customer = {'customer_id': customer_id, 'name': name}
        self.customers.append(customer)
        return True

        

    def add_order(self, order_id: int, customer_id: int, description: str, weight: float) -> bool:
       
        for order in self.orders: 
            if order['order_id'] == order_id or order['customer_id'] == customer_id or weight <= 0:
                return False

        order = {'order_id': order_id,'customer_id': customer_id, 'description': description,'weight': weight,'status':'pending','driver_id':'None'}
        self.orders.append(order)
        return True

    def get_order(self, order_id: int) -> dict | None:
        for order in self.orders:
            if order['order_id'] == order_id:
                return order

        return None

    def update_order_status(self, order_id: int, status: str) -> bool:
        for order in self.orders:
            if order['order_id'] == order_id and (status == 'pending' or status == 'in_transit' or status == 'delivered'):
                order['status'] = status
                return True
                
        return False
                

    def get_customer_orders(self, customer_id: int) -> list[dict]:
        
        customer_orders = []
        found = False
        
        for order in self.orders:
            if order['customer_id'] == customer_id:
                customer_orders.append(order)
                found = True

        for order in customer_orders:
            for order1 in self.orders:
                if order['weight'] == order1['weight']:
                    customer_orders.sort(key = lambda customer_orders: customer_orders['order_id'])
        
        customer_orders.sort(key = lambda customer_order: customer_order['weight'],reverse=True)  
        
        if found == False:
            return []

        return customer_orders

    def get_orders_by_status(self, status: str) -> list[dict]:

        orders_by_status = []
        
        for order in self.orders:
            if order['status'] == status and (status == 'pending' or status == 'in_transit' or status == 'delivered'):
                orders_by_status.append(order)

        orders_by_status.sort(key = lambda order_by_status: order_by_status['order_id'])
        return orders_by_status

    
    def add_driver(self, driver_id: int, name: str) -> bool:
        
        for driver in self.drivers:
            if driver['driver_id'] == driver_id:
                return False
        
        driver = {'driver_id': driver_id,'name': name}
        self.drivers.append(driver)
        return True

    def assign_order(self, order_id: int, driver_id: int) -> bool:

        for order in self.orders:
            for driver in self.drivers:
                if order['order_id'] == order_id and driver['driver_id'] == driver_id:
                    order['driver_id'] = driver_id
                    return True
        return False

    def get_driver_workload(self, driver_id: int) -> int:

        count = 0
        for order in self.orders:
            if order['driver_id'] == driver_id:
                count += 1

        return count

    def get_busiest_drivers(self, n: int) -> list[int]:
        if n <= 0 or n > len(self.drivers):
            return []

        workload = {}
        names = {}

        for driver in self.drivers:
            workload['driver_id'] = 0
            names['driver_name'] = driver['name']

        for order in orders:
            if order['driver_id'] == driver_id and order['status'] != 'delivered':
                workload[order['driver_id']] += 1

        driver_ids = list(workload.keys())
        driver_ids.sort(reverse=True)

        return drivers_ids[:n]
        


    def get_driver_delivery_rate(self, driver_id: int) -> float:

        count = 0
        
        for driver in self.drivers:
            if driver['driver_id'] == driver_id and manager.get_driver_workload(driver_id) > 0:
                for order in self.orders:
                    if order['status'] == 'delivered' and order['driver_id'] == driver_id:
                        count += 1

        delivery_rate = count/manager.get_driver_workload(driver_id)

        return count

    def get_top_drivers(self, n: int) -> list[int]:
        pass
```
