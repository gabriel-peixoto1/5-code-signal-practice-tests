Level 1 — User and Task Management

Create a system capable of managing users and tasks.

Each user must contain:
- user_id
- name

Each task must contain:
- task_id
- title
- user_id
- status
- priority

A task's initial status must be:
- "pending"

add_user(user_id, name)

- Adds a new user.
- If user_id already exists, return False.
- Otherwise, create the user and return True.

```python
users = []
tasks = []
```

```python
def add_user(user_id,name):    # edge case: user_id negativo ou menor que zero? 
    for user in users:
        if user['user_id'] == user_id:
            print('já existe.')
            return False;
    
    user = {}
    user['user_id'] = user_id
    user['name'] = name
    users.append(user)

    return True;
```

get_user(user_id)

- Returns the user associated with user_id.
- If the user does not exist, return None.

```python
def get_user(user_id):    # edge case: user_id não existe. 
    for user in users:
        if user['user_id'] == user_id:
            return user;
    return None;
    
```

add_task(task_id, title, user_id, priority)

- Creates a task assigned to an existing user.
- If task_id already exists, return False.
- If user_id does not exist, return False.
- If priority is less than 1, return False.
- Otherwise, create the task with status "pending" and return True.

```python
def add_task(task_id, title, user_id, priority):

    if priority < 1:
        return False;
    
    for task in tasks:
        if task['task_id'] == task_id:
            return False;

    found = False
    
    for user in users:
        if user['user_id'] == user_id:
            found = True;

    if found == False:
        return False; 

    task = {}
    task['task_id'] = task_id
    task['title'] = title
    task['user_id'] = user_id
    task['priority'] = priority
    task['status'] = 'pending'
    
    tasks.append(task)
    return True;
```

get_task(task_id)

- Returns the task associated with task_id.
- If the task does not exist, return None.

```python
def get_task(task_id):
    for task in tasks:
        if task['task_id'] == task_id:
            return task
    return None;
```

Level 2 — Task Operations and Filtering

Extend the existing system.

update_task_status(task_id, status)

- Updates the status of a task.

Allowed statuses are:

- "pending"
- "in_progress"
- "completed"

Return:

- False if the task does not exist.
- False if the status is invalid.
- True otherwise.

```python
def update_task_status(task_id,status): # edge cases: task_id or status doesn't exist. 
    for task in tasks:
        if task['task_id'] == task_id:
            if status == 'pending' or status == 'in_progress' or status == 'completed':
                task['status'] = status
                return True;
    return False;
```

delete_task(task_id)

- Deletes a task.

Return:

- False if the task does not exist.
- True if the task was successfully deleted.

```python
def delete_task(task_id):
    for i,task in enumerate(tasks):      # a função enumerate permite que eu enumere cada indice do vetor tasks, e que para cada task procurada, eu tenha o indice correspondente junto. 
        if task['task_id'] == task_id:
            del tasks[i]
            return True

    return False
```

get_user_tasks(user_id)

Returns all tasks assigned to the specified user.

- Return [] if the user does not exist.
- Return [] if the user has no tasks.
- The result must be sorted by:
- priority descending;
- task ID ascending in case of a tie.

```python
def get_user_tasks(user_id):
    
    tasks_by_user = []
    found = False
    
    for task in tasks:
        if task['user_id'] == user_id:
            tasks_by_user.append(task)
            found = True
                           
    if found == False:
        return []

    for task in tasks_by_user:
        for task1 in tasks_by_user:
            if task['priority'] == task1['priority']:
                tasks_by_user.sort(key = lambda tasks_by_user: tasks_by_user['task_id'])   # percorrendo o vetor duas vezes e analisando: se a prioridade entre as duas for igual, então ordenar por task_id.
                
    tasks_by_user.sort(key = lambda tasks_by_user: tasks_by_user['priority'],reverse=True)   # reverse=True transforma em decrescente.
    
    return tasks_by_user
```

get_tasks_by_status(status)

Returns all tasks with the specified status.

- If the status is invalid, return [].
- Sort the result by task_id ascending.

```python
def get_tasks_by_status(status):
   
    tasks_by_status = []
    found = False
    
    for task in tasks:
        if task['status'] == status:
            tasks_by_status.append(task)
            found = True
                           
    if found == False:
        return []
                
    tasks_by_status.sort(key = lambda tasks_by_status: tasks_by_status['task_id'])
    
    return tasks_by_status
```

Level 3 — Workload and Task Reassignment

The system must now support reassignment and workload statistics.

reassign_task(task_id, new_user_id)

Moves a task from its current user to another user.

Return:

- False if the task does not exist.
- False if new_user_id does not exist.
- True otherwise.

The task's status and priority must remain unchanged.

```python
def reassign_task(task_id, new_user_id):
    
    for user in users:
        for task in tasks:
            if user['user_id'] == new_user_id and task['task_id'] == task_id:
                task['user_id'] = new_user_id
                return True
    
    return False;
            
```

get_user_workload(user_id)

Returns the number of non-completed tasks assigned to the user.

- Completed tasks must not be counted.
- Return: 0  if  the user does not exist or has no active tasks.

```python
def get_user_workload(user_id: int) -> int:
    
    count = 0
    
    for task in tasks:
        if task['user_id'] == user_id and len(get_user_tasks(user_id)) >= 1 and task['status'] != 'completed':
            count = count + 1
        
    if count > 0:
        return count
    
    return 0
```

get_user_priority_score(user_id)

- Calculate the user's priority score using only non-completed tasks.

Each task contributes:

- priority × 1
- to the user's score.

For example:

- Task A → priority 3
- Task B → priority 2
- Task C → priority 5 (completed)

The score is:

- 3 + 2 = 5
- because completed tasks are ignored.
- Return 0 if the user does not exist or has no active tasks.

```python
def get_user_priority_score(user_id):

    priority_score = 0
    
    for task in tasks:
        if task['user_id'] == user_id and task['status'] != 'completed' and len(get_user_tasks(user_id)) >= 1:
            priority_score = priority_score + task['priority']

    if priority_score > 0:
        return priority_score
    
    return 0
```

get_busiest_users(n)

- Return the top n users according to their number of non-completed tasks.

Sort by:

- number of active tasks descending;
- user name alphabetically in case of a tie.

- Return the result as a list of user IDs.

Rules:

- If n <= 0, return [].
- If n is greater than the number of users, return all users.
- Users with zero active tasks are still valid users and may appear in the result.

```python
def get_busiest_users(n):
    if n <= 0:
        return []

    workload = {}
    names = {}

    # Inicializa todos os usuários
    for user in users:
        workload[user['user_id']] = 0
        names[user['user_id']] = user['name']

    # Conta as tarefas ativas de cada usuário
    for task in tasks:
        if task['status'] != 'completed':
            workload[task['user_id']] += 1

    # Cria uma lista com os IDs dos usuários
    user_ids = list(workload.keys())

    # Ordena:
    # 1. maior workload primeiro
    # 2. em caso de empate, nome alfabeticamente
    user_ids.sort(key=lambda uid: (-workload[uid], names[uid]))

    # Retorna os primeiros n
    return user_ids[:n]
```
