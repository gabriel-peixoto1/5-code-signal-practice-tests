Level 1 - Produtct Management:

Each product has:

a unique product_id;
a name;
a price;
a stock quantity.

Implement the following operations:

* add_product(product_id, name, price, stock)

- Adds a new product to the system.
- If a product with the same product_id already exists, the operation must fail and return False.
- Otherwise, add the product and return True.

* get_product(product_id)

- Returns the product associated with the given ID.
- If the product does not exist, return None.

* update_stock(product_id, quantity)

- Changes the stock of an existing product by adding quantity to its current stock.

```python
storage = []        # every time this cell is run, the storage vector is emptied. Careful.

def add_product(id,name,price,stock):    #should the function always have the same signature than asked in the exercise, including the same name of variables?
    for product in storage:
        if product['product_id'] == id:
            return False;
    product = {}
    product['product_id'] = id
    product['name'] = name
    product['price'] = price
    product['stock'] = stock
    storage.append(product)
    return True;               # when do I put the ';' ?
     

def get_product(id):
    for product in storage:
        if product['product_id'] == id:
            return product;
    return None;              # why the return works outside the loop, but not in a 'else' inside the loop?

def update_stock(id,quantity):
    for product in storage:
        if product['product_id'] == id and quantity > 0:
            product['stock'] = product['stock'] + quantity
            return True;
    return False;
        
```

Level 2 - Orders: 

The store now needs to process customer orders.

An order contains:

- a unique order_id;
- a customer;
- a collection of products and quantities;
- the total order value.

Implement:

create_order(order_id, customer, items)

- obs: 'items' contains product IDs and quantities requested by the customer.

The operation must fail and return False if:
- the order ID already exists;
- any requested product does not exist;
- any requested quantity is less than or equal to zero;
- any product does not have enough stock.

If the order is successfully created:

- decrease the stock of every requested product;
- calculate the total order value;
- store the order;
- return True.

he total must be calculated using the price of each product at the moment the order is created.

```python
orders = []

def create_order(order_id,customer,items):     # how can I descbre 'items' as a dictionary?
    
    for order in orders:
        if order['order_id'] == order_id:
            return False;
    
    for key, value in items.items():    # usando a função items() para avaliar, individualmente, as chaves e valores dentro do dicionário 'items'. 
        
        found = False        # usarei essa variável auxiliar. 

        for product in storage:      
            if product['product_id'] == key:   # para cada produto em storage, só irei avaliar se as id's forem iguais. 
                found = True;           # se as id's forem iguais, então a id correta foi 'encontrada'. 

                if value <= 0 or product['stock'] < value:      # agora, chegando se o produto da id encontrada atende as caractetísticas pedidas.
                    return False;      
        
    if found == False:     # se não encontrado, return False.
        return False;
                
    
    order = {}                         # se tudo certo com as condições, vamos adicionar essa ordem às ordens. 
    order['order_id'] = order_id          
    order['customer'] = customer
    order['items'] = items      # 'items' will be added as a dictionary, linke {67:3,31:4} means 3 of the id 67 and 4 of the id 31.
    order['total_value'] = 0    # o valor total da nota começa em zero.
    order['canceled'] = False
    
    for code,quantity in items.items():          
        for product in storage:
            if code == product['product_id']:
                order['total_value'] = order['total_value'] + (product['price'] * quantity)    # o valor da nota se atualiza a cada loop, ou seja, à cada produto do pedido items().
                product['stock'] = product['stock'] - quantity  

    orders.append(order)
    return True;
    
```

Implement:

```python
def get_order(order_id):
    
    for order in orders:
        if order_id == order['order_id']:
            return order;
    
    return None;
```

cancel_order(order_id)

- Cancels an existing order.

- If the order has already been cancelled, return False.

Otherwise:

- mark the order as cancelled;
- restore the ordered quantities to the corresponding products' stock;
- return True.

A cancelled order must not be counted as an active order.

```python
def cancel_order(order_id):
    for order in orders:
        if order['order_id'] == order_id:
            if order['canceled'] == False:
                order['canceled'] = True;
                for product in storage:
                    if order['order_id'] == product['product_id']:
                        for code, quantity in order['items'].items():
                            product['stock'] = product['stock'] + quantity
                            order['total_value'] = order['total_value'] - (product['price']*quantity)
                return True;

    return False;

# A cancelled order must not be counted as an active order (???).
```
