Crie uma estrutura para armazenar os livros da biblioteca.

Cada livro deve possuir:

id
titulo
autor
disponivel

Inicialmente, a biblioteca estará vazia.

Implemente as seguintes funções:

adicionar_livro(livros, id, titulo, autor)
buscar_livro(livros, id)

```python
biblioteca = []  # será um vetor de dicionários, livros.

def adicionar_livro(biblioteca,id,titulo,autor):
    for livro in biblioteca:
        if livro["id"] == id:
            print("Livro já adicionado.")
            return;
    livro = {}
    livro['id'] = id
    livro['titulo'] = titulo
    livro['autor'] = autor
    livro['disponivel'] = True
    biblioteca.append(livro)

def buscar_livro(biblioteca,id):
    for livro in biblioteca:
        if livro["id"] == id:
            return livro
            
    return None;
    
```

```python
adicionar_livro(biblioteca,31,"Superman","Mark Millar")
adicionar_livro(biblioteca,19,"Sandman","Neil Gaiman")
adicionar_livro(biblioteca,11,"Hellblazer","Alan Moore")
adicionar_livro(biblioteca,67,"Turma da Mônica","Maurício de Souza")
adicionar_livro(biblioteca,99,"The boys","Eric Kripke")
adicionar_livro(biblioteca,200,"Mathematical Way of Thinking","Jordan Ellenberg")
adicionar_livro(biblioteca,999,"Animal Man","Alan Moore")
adicionar_livro(biblioteca,787,"Lucifer","Neil Gaiman")
adicionar_livro(biblioteca,434,"Watchemn","Alan Moore")
```

Agora a biblioteca precisa permitir que livros sejam emprestados e devolvidos.

Crie as funções:

emprestar_livro(livros, id)
devolver_livro(livros, id)

emprestar_livro: 

- Ao receber um id:

-Se o livro não existir → retorne False.
-Se o livro já estiver emprestado → retorne False.
-Caso contrário:
-altere disponivel para False;
-retorne True.

devolver_livro:

-Se o livro não existir → False.
-Se o livro já estiver disponível → False.
-Caso contrário:
-altere disponivel para True;
-retorne True.

```python
def emprestar_livro(biblioteca,id):
    for livro in biblioteca:
        if livro["id"] == id:
            if livro["disponivel"] == False:
                return False;
            else:
                livro['disponivel'] = False
                return True;
                

    return False;
```

```python
def devolver_livro(biblioteca,id):
    for livro in biblioteca:
        if livro["id"] == id:
            if livro["disponivel"] == True:
                return False;
            else:
                livro['disponivel'] = True
                return True;
                

    return False
```

A biblioteca agora quer oferecer uma função de pesquisa.

Crie:
- buscar_por_autor(livros, autor)
- A função deve retornar uma lista contendo todos os livros daquele autor.
- A pesquisa deve ser feita por correspondência exata do nome do autor.

Agora crie:

- listar_livros(livros)
- Ela deve retornar todos os livros disponíveis, ordenados alfabeticamente pelo título.

```python
def buscar_por_autor(biblioteca,autor):
    livros_autor = []
    for livro in biblioteca:
        if livro['autor'] == autor:
            livros_autor.append(livro)
    return livros_autor
```

```python
buscar_por_autor(biblioteca,"Alan Moore")
```

```python
def listar_livros(biblioteca):
    disponiveis = []
    for livro in biblioteca:
        if livro['disponivel'] == True:
            disponiveis.append(livro)
                
    disponiveis.sort(key = lambda disponiveis: disponiveis['titulo'])        
    return disponiveis
```

```python
def emprestar_livro(biblioteca,id,usuario):
    for livro in biblioteca:
        if livro["id"] == id:
            if livro["disponivel"] == False:
                return False;
            else:
                livro['disponivel'] = False
                livro['emprestado_para'] = usuario
                return True;
                

    return False;
    
def devolver_livro(biblioteca,id):
    for livro in biblioteca:
        if livro["id"] == id:
            if livro["disponivel"] == True:
                return False;
            else:
                livro['disponivel'] = True
                livro['emprestado_para'] = None
                return True;
                

    return False    

def livros_do_usuario(biblioteca, usuario):
    resultado = []

    for livro in biblioteca:
        if livro["emprestado_para"] == usuario:
            resultado.append(livro)

    return resultado
```

```python
livros_do_usuario(biblioteca, 'Gabriel')
```
